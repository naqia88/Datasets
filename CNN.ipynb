# CUSTOM CNN

```python
# ============================================================
# TRADITIONAL CNN - PlantDoc (Paper Ready)
# ============================================================

# STEP 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# STEP 2: Imports
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# STEP 3: Paths
TRAIN_DIR = '/content/PlantDoc-Dataset/train'
TEST_DIR  = '/content/PlantDoc-Dataset/test'

# STEP 4: Data Loading
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    horizontal_flip=True,
    rotation_range=15,
    zoom_range=0.1
)

test_datagen = ImageDataGenerator(rescale=1./255)

train_data = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

test_data = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False  # important for classification report
)

NUM_CLASSES = len(train_data.class_indices)
print(f"Number of classes: {NUM_CLASSES}")

# STEP 5: Build Traditional CNN (without BatchNormalization)
model = models.Sequential([
    # Block 1
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(224,224,3)),
    layers.MaxPooling2D(2,2),

    # Block 2
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    # Block 3
    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    # Block 4
    layers.Conv2D(256, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    # Classifier
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

model.summary()

# STEP 6: Compile
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# STEP 7: Train
early_stop = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)

history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=30,
    callbacks=[early_stop],
    verbose=1
)

# STEP 8: Plot Accuracy & Loss
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,4))

ax1.plot(history.history['accuracy'], label='Train Accuracy')
ax1.plot(history.history['val_accuracy'], label='Val Accuracy')
ax1.set_title('CNN Accuracy')
ax1.legend()

ax2.plot(history.history['loss'], label='Train Loss')
ax2.plot(history.history['val_loss'], label='Val Loss')
ax2.set_title('CNN Loss')
ax2.legend()

plt.savefig('/content/drive/MyDrive/cnn_accuracy_loss.png')
plt.show()

# STEP 9: Evaluate
loss, accuracy = model.evaluate(test_data)
print(f"\nTest Accuracy: {accuracy*100:.2f}%")
print(f"Test Loss: {loss:.4f}")

# STEP 10: Classification Report
y_pred = model.predict(test_data)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = test_data.classes
class_names = list(test_data.class_indices.keys())

print("\nClassification Report:")
print(classification_report(y_true, y_pred_classes, target_names=class_names))

# STEP 11: Confusion Matrix
cm = confusion_matrix(y_true, y_pred_classes)
plt.figure(figsize=(20,16))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix - Traditional CNN')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/cnn_confusion_matrix.png')
plt.show()

# STEP 12: Save Model
model.save('/content/drive/MyDrive/traditional_cnn_plantdoc.keras')
print("Model saved!")
```

# VGG-16

```python
# ============================================================
# VGG-16 - PlantDoc (Transfer Learning)
# ============================================================

# STEP 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# STEP 2: Imports
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# STEP 3: Paths
TRAIN_DIR = '/content/PlantDoc-Dataset/train'
TEST_DIR  = '/content/PlantDoc-Dataset/test'

# STEP 4: Data Loading
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    horizontal_flip=True,
    rotation_range=15,
    zoom_range=0.1
)

test_datagen = ImageDataGenerator(rescale=1./255)

train_data = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

test_data = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

NUM_CLASSES = len(train_data.class_indices)
print(f"Number of classes: {NUM_CLASSES}")

# STEP 5: Load VGG-16 Base Model
base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze base model layers
base_model.trainable = False

# STEP 6: Add Custom Classification Head
x = base_model.output
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.5)(x)
output = layers.Dense(NUM_CLASSES, activation='softmax')(x)

model = tf.keras.Model(inputs=base_model.input, outputs=output)

model.summary()

# STEP 7: Compile
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# STEP 8: Train (Phase 1 - Frozen base)
early_stop = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)

history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=30,
    callbacks=[early_stop],
    verbose=1
)

# STEP 9: Fine Tuning (Phase 2 - Unfreeze last few layers)
base_model.trainable = True

# Freeze all except last 4 layers
for layer in base_model.layers[:-4]:
    layer.trainable = False

# Recompile with lower learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history_fine = model.fit(
    train_data,
    validation_data=test_data,
    epochs=20,
    callbacks=[early_stop],
    verbose=1
)

# STEP 10: Plot Accuracy & Loss
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,4))

ax1.plot(history.history['accuracy'] + history_fine.history['accuracy'], label='Train Accuracy')
ax1.plot(history.history['val_accuracy'] + history_fine.history['val_accuracy'], label='Val Accuracy')
ax1.set_title('VGG-16 Accuracy')
ax1.legend()

ax2.plot(history.history['loss'] + history_fine.history['loss'], label='Train Loss')
ax2.plot(history.history['val_loss'] + history_fine.history['val_loss'], label='Val Loss')
ax2.set_title('VGG-16 Loss')
ax2.legend()

plt.savefig('/content/drive/MyDrive/vgg16_accuracy_loss.png')
plt.show()

# STEP 11: Evaluate
loss, accuracy = model.evaluate(test_data)
print(f"\nTest Accuracy: {accuracy*100:.2f}%")
print(f"Test Loss: {loss:.4f}")

# STEP 12: Classification Report
y_pred = model.predict(test_data)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = test_data.classes
class_names = list(test_data.class_indices.keys())

print("\nClassification Report:")
print(classification_report(y_true, y_pred_classes, target_names=class_names))

# STEP 13: Confusion Matrix
cm = confusion_matrix(y_true, y_pred_classes)
plt.figure(figsize=(20,16))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix - VGG-16')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/vgg16_confusion_matrix.png')
plt.show()

# STEP 14: Save Model
model.save('/content/drive/MyDrive/vgg16_plantdoc.keras')
print("Model saved!")
```

# EfficientNet

```python
# ============================================================
# EfficientNetB0 - PlantDoc (Transfer Learning)
# ============================================================

# STEP 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# STEP 2: Imports
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetV2S
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# STEP 3: Paths
TRAIN_DIR = '/content/PlantDoc-Dataset/train'
TEST_DIR  = '/content/PlantDoc-Dataset/test'

# STEP 4: Data Loading
# NOTE: EfficientNet has its own preprocessing built in
# so we do NOT rescale — just pass raw 0-255 pixel values
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    horizontal_flip=True,
    rotation_range=15,
    zoom_range=0.1
)

test_datagen = ImageDataGenerator()

train_data = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

test_data = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

NUM_CLASSES = len(train_data.class_indices)
print(f"Number of classes: {NUM_CLASSES}")

# STEP 5: Load EfficientNetB0 Base Model
base_model = EfficientNetV2S(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze base model layers
base_model.trainable = False

# STEP 6: Add Custom Classification Head
x = base_model.output
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.5)(x)
output = layers.Dense(NUM_CLASSES, activation='softmax')(x)

model = tf.keras.Model(inputs=base_model.input, outputs=output)

model.summary()

# STEP 7: Compile
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# STEP 8: Train Phase 1 - Frozen base
early_stop = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)

history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=30,
    callbacks=[early_stop],
    verbose=1
)

# STEP 9: Fine Tuning Phase 2
base_model.trainable = True

for layer in base_model.layers[:-4]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history_fine = model.fit(
    train_data,
    validation_data=test_data,
    epochs=20,
    callbacks=[early_stop],
    verbose=1
)

# STEP 10: Plot Accuracy & Loss
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,4))

ax1.plot(history.history['accuracy'] + history_fine.history['accuracy'], label='Train Accuracy')
ax1.plot(history.history['val_accuracy'] + history_fine.history['val_accuracy'], label='Val Accuracy')
ax1.set_title('EfficientNetB0 Accuracy')
ax1.legend()

ax2.plot(history.history['loss'] + history_fine.history['loss'], label='Train Loss')
ax2.plot(history.history['val_loss'] + history_fine.history['val_loss'], label='Val Loss')
ax2.set_title('EfficientNetB0 Loss')
ax2.legend()

plt.savefig('/content/drive/MyDrive/efficientnet_accuracy_loss.png')
plt.show()

# STEP 11: Evaluate
loss, accuracy = model.evaluate(test_data)
print(f"\nTest Accuracy: {accuracy*100:.2f}%")
print(f"Test Loss: {loss:.4f}")

# STEP 12: Classification Report
y_pred = model.predict(test_data)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = test_data.classes
class_names = list(test_data.class_indices.keys())

print("\nClassification Report:")
print(classification_report(y_true, y_pred_classes, target_names=class_names))

# STEP 13: Confusion Matrix
cm = confusion_matrix(y_true, y_pred_classes)
plt.figure(figsize=(20,16))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix - EfficientNetB0')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/efficientnet_confusion_matrix.png')
plt.show()

# STEP 14: Save Model
model.save('/content/drive/MyDrive/efficientnet_plantdoc.keras')
print("Model saved!")
```

# Swin Transformer

```python
# ============================================================
# Swin Transformer - PlantDoc (Transfer Learning)
# ============================================================

# STEP 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# STEP 2: Install required library
!pip install transformers timm -q

# STEP 3: Imports
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from transformers import SwinForImageClassification, AutoFeatureExtractor
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm

# STEP 4: Paths
TRAIN_DIR = '/content/PlantDoc-Dataset/train'
TEST_DIR  = '/content/PlantDoc-Dataset/test'

# STEP 5: Data Loading
# Swin uses 224x224
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

transform_test = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

train_dataset = datasets.ImageFolder(TRAIN_DIR, transform=transform_train)
test_dataset  = datasets.ImageFolder(TEST_DIR,  transform=transform_test)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True,  num_workers=2)
test_loader  = DataLoader(test_dataset,  batch_size=32, shuffle=False, num_workers=2)

NUM_CLASSES = len(train_dataset.classes)
print(f"Number of classes: {NUM_CLASSES}")
print(f"Classes: {train_dataset.classes}")

# STEP 6: Load Swin Transformer
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

model = SwinForImageClassification.from_pretrained(
    'microsoft/swin-tiny-patch4-window7-224',
    num_labels=NUM_CLASSES,
    ignore_mismatched_sizes=True
)
model = model.to(device)

# STEP 7: Freeze base, only train classifier
for param in model.parameters():
    param.requires_grad = False

# Unfreeze classifier head
for param in model.classifier.parameters():
    param.requires_grad = True

# STEP 8: Optimizer and Loss
optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4
)
criterion = nn.CrossEntropyLoss()

# STEP 9: Training Function
def train_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss, correct, total = 0, 0, 0
    for images, labels in tqdm(loader, desc='Training'):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images).logits
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        _, predicted = outputs.max(1)
        correct += predicted.eq(labels).sum().item()
        total += labels.size(0)
    return total_loss/len(loader), correct/total

# STEP 10: Evaluation Function
def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images).logits
            loss = criterion(outputs, labels)
            total_loss += loss.item()
            _, predicted = outputs.max(1)
            correct += predicted.eq(labels).sum().item()
            total += labels.size(0)
    return total_loss/len(loader), correct/total

# STEP 11: Phase 1 Training - Frozen base
print("\n--- Phase 1: Training classifier head ---")
train_losses, train_accs = [], []
val_losses, val_accs = [], []
best_val_acc = 0
patience_counter = 0
PATIENCE = 5

for epoch in range(30):
    train_loss, train_acc = train_epoch(model, train_loader, optimizer, criterion, device)
    val_loss, val_acc = evaluate(model, test_loader, criterion, device)

    train_losses.append(train_loss)
    train_accs.append(train_acc)
    val_losses.append(val_loss)
    val_accs.append(val_acc)

    print(f"Epoch {epoch+1}/30 | Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}% | Val Loss: {val_loss:.4f} | Val Acc: {val_acc*100:.2f}%")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), '/content/best_swin.pth')
        patience_counter = 0
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print("Early stopping!")
            break

# STEP 12: Phase 2 Fine Tuning - Unfreeze last few layers
print("\n--- Phase 2: Fine tuning ---")
model.load_state_dict(torch.load('/content/best_swin.pth'))

# Unfreeze last transformer layer
for param in model.swin.layernorm.parameters():
    param.requires_grad = True
for param in model.swin.encoder.layers[-1].parameters():
    param.requires_grad = True

optimizer_fine = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-5
)

best_val_acc_fine = 0
patience_counter = 0

for epoch in range(20):
    train_loss, train_acc = train_epoch(model, train_loader, optimizer_fine, criterion, device)
    val_loss, val_acc = evaluate(model, test_loader, criterion, device)

    train_losses.append(train_loss)
    train_accs.append(train_acc)
    val_losses.append(val_loss)
    val_accs.append(val_acc)

    print(f"Epoch {epoch+1}/20 | Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}% | Val Loss: {val_loss:.4f} | Val Acc: {val_acc*100:.2f}%")

    if val_acc > best_val_acc_fine:
        best_val_acc_fine = val_acc
        torch.save(model.state_dict(), '/content/best_swin_fine.pth')
        patience_counter = 0
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print("Early stopping!")
            break

# STEP 13: Load best model and evaluate
model.load_state_dict(torch.load('/content/best_swin_fine.pth'))

# STEP 14: Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,4))
ax1.plot(train_accs, label='Train Accuracy')
ax1.plot(val_accs, label='Val Accuracy')
ax1.set_title('Swin Transformer Accuracy')
ax1.legend()
ax2.plot(train_losses, label='Train Loss')
ax2.plot(val_losses, label='Val Loss')
ax2.set_title('Swin Transformer Loss')
ax2.legend()
plt.savefig('/content/drive/MyDrive/swin_accuracy_loss.png')
plt.show()

# STEP 15: Classification Report
model.eval()
all_preds, all_labels = [], []
with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        outputs = model(images).logits
        _, predicted = outputs.max(1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.numpy())

print(f"\nTest Accuracy: {(np.array(all_preds)==np.array(all_labels)).mean()*100:.2f}%")
print("\nClassification Report:")
print(classification_report(all_labels, all_preds, target_names=train_dataset.classes))

# STEP 16: Confusion Matrix
cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(20,16))
sns.heatmap(cm, annot=True, fmt='d',
            xticklabels=train_dataset.classes,
            yticklabels=train_dataset.classes)
plt.title('Confusion Matrix - Swin Transformer')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/swin_confusion_matrix.png')
plt.show()

# STEP 17: Save Model
torch.save(model.state_dict(), '/content/drive/MyDrive/swin_plantdoc.pth')
print("Model saved!")
```

```python
# ============================================================
# ViT (Vision Transformer) - PlantDoc (Transfer Learning)
# ============================================================

# STEP 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# STEP 2: Install required library
!pip install transformers timm -q

# STEP 3: Imports
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from transformers import ViTForImageClassification
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm

# STEP 4: Paths
TRAIN_DIR = '/content/PlantDoc-Dataset/train'
TEST_DIR  = '/content/PlantDoc-Dataset/test'

# STEP 5: Data Loading
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

transform_test = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

train_dataset = datasets.ImageFolder(TRAIN_DIR, transform=transform_train)
test_dataset  = datasets.ImageFolder(TEST_DIR,  transform=transform_test)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True,  num_workers=2)
test_loader  = DataLoader(test_dataset,  batch_size=32, shuffle=False, num_workers=2)

NUM_CLASSES = len(train_dataset.classes)
print(f"Number of classes: {NUM_CLASSES}")

# STEP 6: Load ViT
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

model = ViTForImageClassification.from_pretrained(
    'google/vit-base-patch16-224',
    num_labels=NUM_CLASSES,
    ignore_mismatched_sizes=True
)
model = model.to(device)

# STEP 7: Freeze base, only train classifier
for param in model.parameters():
    param.requires_grad = False

for param in model.classifier.parameters():
    param.requires_grad = True

# STEP 8: Optimizer and Loss
optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4
)
criterion = nn.CrossEntropyLoss()

# STEP 9: Training Function
def train_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss, correct, total = 0, 0, 0
    for images, labels in tqdm(loader, desc='Training'):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images).logits
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        _, predicted = outputs.max(1)
        correct += predicted.eq(labels).sum().item()
        total += labels.size(0)
    return total_loss/len(loader), correct/total

# STEP 10: Evaluation Function
def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images).logits
            loss = criterion(outputs, labels)
            total_loss += loss.item()
            _, predicted = outputs.max(1)
            correct += predicted.eq(labels).sum().item()
            total += labels.size(0)
    return total_loss/len(loader), correct/total

# STEP 11: Phase 1 Training - Frozen base
print("\n--- Phase 1: Training classifier head ---")
train_losses, train_accs = [], []
val_losses, val_accs = [], []
best_val_acc = 0
patience_counter = 0
PATIENCE = 5

for epoch in range(30):
    train_loss, train_acc = train_epoch(model, train_loader, optimizer, criterion, device)
    val_loss, val_acc = evaluate(model, test_loader, criterion, device)

    train_losses.append(train_loss)
    train_accs.append(train_acc)
    val_losses.append(val_loss)
    val_accs.append(val_acc)

    print(f"Epoch {epoch+1}/30 | Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}% | Val Loss: {val_loss:.4f} | Val Acc: {val_acc*100:.2f}%")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), '/content/best_vit.pth')
        patience_counter = 0
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print("Early stopping!")
            break

# STEP 12: Phase 2 Fine Tuning
print("\n--- Phase 2: Fine tuning ---")
model.load_state_dict(torch.load('/content/best_vit.pth'))

# ============================================================
# STEP 12: Phase 2 - Fine Tuning
# ============================================================

print("\n--- Phase 2: Fine tuning ---")

# Load the best model from Phase 1
model.load_state_dict(torch.load('/content/best_vit.pth'))

# ------------------------------------------------------------
# Freeze everything first
# ------------------------------------------------------------

for param in model.parameters():
    param.requires_grad = False

# ------------------------------------------------------------
# Unfreeze classifier
# ------------------------------------------------------------

for param in model.classifier.parameters():
    param.requires_grad = True

# ------------------------------------------------------------
# Unfreeze final LayerNorm
# ------------------------------------------------------------

for param in model.vit.layernorm.parameters():
    param.requires_grad = True

# ------------------------------------------------------------
# Unfreeze the LAST Transformer block
# ViT has 12 blocks: 0-11
# ------------------------------------------------------------

for param in model.vit.layers[-1].parameters():
    param.requires_grad = True

# ------------------------------------------------------------
# Check which parameters are trainable
# ------------------------------------------------------------

print("\nTrainable parameters:")

trainable_params = 0
total_params = 0

for name, param in model.named_parameters():
    total_params += param.numel()

    if param.requires_grad:
        trainable_params += param.numel()
        print(name)

print(f"\nTrainable parameters: {trainable_params:,}")
print(f"Total parameters: {total_params:,}")
print(
    f"Percentage trainable: "
    f"{100 * trainable_params / total_params:.2f}%"
)

# ------------------------------------------------------------
# Optimizer
# ------------------------------------------------------------

optimizer_fine = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-5
)

criterion = nn.CrossEntropyLoss()

# ------------------------------------------------------------
# Fine-tuning
# ------------------------------------------------------------

best_val_acc_fine = 0
patience_counter = 0
PATIENCE = 5

for epoch in range(20):

    train_loss, train_acc = train_epoch(
        model,
        train_loader,
        optimizer_fine,
        criterion,
        device
    )

    val_loss, val_acc = evaluate(
        model,
        test_loader,
        criterion,
        device
    )

    train_losses.append(train_loss)
    train_accs.append(train_acc)
    val_losses.append(val_loss)
    val_accs.append(val_acc)

    print(
        f"Fine-tuning Epoch {epoch+1}/20 | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc*100:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc*100:.2f}%"
    )

    # Save best fine-tuned model
    if val_acc > best_val_acc_fine:

        best_val_acc_fine = val_acc

        torch.save(
            model.state_dict(),
            '/content/best_vit_fine.pth'
        )

        patience_counter = 0

    else:

        patience_counter += 1

        if patience_counter >= PATIENCE:

            print("Early stopping!")
            break

print(
    f"\nBest Phase 2 Validation Accuracy: "
    f"{best_val_acc_fine*100:.2f}%"
)

# ============================================================
# STEP 13: Load Best Fine-Tuned Model
# ============================================================

model.load_state_dict(
    torch.load('/content/best_vit_fine.pth')
)

model = model.to(device)

print("Best fine-tuned model loaded successfully.")

# STEP 14: Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,4))
ax1.plot(train_accs, label='Train Accuracy')
ax1.plot(val_accs, label='Val Accuracy')
ax1.set_title('ViT Accuracy')
ax1.legend()
ax2.plot(train_losses, label='Train Loss')
ax2.plot(val_losses, label='Val Loss')
ax2.set_title('ViT Loss')
ax2.legend()
plt.savefig('/content/drive/MyDrive/vit_accuracy_loss.png')
plt.show()

# STEP 15: Classification Report
model.eval()
all_preds, all_labels = [], []
with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        outputs = model(images).logits
        _, predicted = outputs.max(1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.numpy())

print(f"\nTest Accuracy: {(np.array(all_preds)==np.array(all_labels)).mean()*100:.2f}%")
print("\nClassification Report:")
print(classification_report(all_labels, all_preds, target_names=train_dataset.classes))

# STEP 16: Confusion Matrix
cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(20,16))
sns.heatmap(cm, annot=True, fmt='d',
            xticklabels=train_dataset.classes,
            yticklabels=train_dataset.classes)
plt.title('Confusion Matrix - ViT')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/vit_confusion_matrix.png')
plt.show()

# STEP 17: Save Model
torch.save(model.state_dict(), '/content/drive/MyDrive/vit_plantdoc.pth')
print("Model saved!")
```

# ConvNeXt

```python
# ============================================================
# ConvNeXt-Tiny - PlantDoc (Transfer Learning)
# Complete Training Pipeline
# ============================================================

# ============================================================
# STEP 1: Mount Google Drive
# ============================================================

from google.colab import drive

drive.mount('/content/drive', force_remount=True)


# ============================================================
# STEP 2: Imports
# ============================================================

import torch
import torch.nn as nn

from torch.utils.data import DataLoader

from torchvision import datasets, transforms
from torchvision.models import convnext_tiny, ConvNeXt_Tiny_Weights

import numpy as np

from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

import matplotlib.pyplot as plt
import seaborn as sns

from tqdm import tqdm


# ============================================================
# STEP 3: Paths
# ============================================================

TRAIN_DIR = '/content/PlantDoc-Dataset/train'
TEST_DIR  = '/content/PlantDoc-Dataset/test'


# ============================================================
# STEP 4: Data Transformations
# ============================================================

transform_train = transforms.Compose([

    transforms.Resize((224, 224)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomRotation(15),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


transform_test = transforms.Compose([

    transforms.Resize((224, 224)),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ============================================================
# STEP 5: Load Dataset
# ============================================================

train_dataset = datasets.ImageFolder(
    TRAIN_DIR,
    transform=transform_train
)

test_dataset = datasets.ImageFolder(
    TEST_DIR,
    transform=transform_test
)


train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=2
)


test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2
)


NUM_CLASSES = len(train_dataset.classes)

print(f"Number of classes: {NUM_CLASSES}")

print("\nClasses:")

for i, class_name in enumerate(train_dataset.classes):

    print(f"{i}: {class_name}")


# ============================================================
# STEP 6: Device
# ============================================================

device = torch.device(
    'cuda' if torch.cuda.is_available() else 'cpu'
)

print(f"\nUsing device: {device}")


# ============================================================
# STEP 7: Load Pretrained ConvNeXt-Tiny
# ============================================================

print("\nLoading pretrained ConvNeXt-Tiny...")

weights = ConvNeXt_Tiny_Weights.DEFAULT

model = convnext_tiny(
    weights=weights
)


# ============================================================
# STEP 8: Replace Classifier
# ============================================================

# Original ConvNeXt classifier:
#
# model.classifier[0] -> LayerNorm
# model.classifier[1] -> Flatten
# model.classifier[2] -> Linear
#
# Replace the final Linear layer
# with one matching PlantDoc classes.

in_features = model.classifier[2].in_features

model.classifier[2] = nn.Linear(
    in_features,
    NUM_CLASSES
)


model = model.to(device)


print(
    f"\nClassifier replaced successfully."
)

print(
    f"Input features: {in_features}"
)

print(
    f"Output classes: {NUM_CLASSES}"
)


# ============================================================
# STEP 9: PHASE 1
# Freeze ConvNeXt Backbone
# Train Only Classifier
# ============================================================

print("\n")
print("=" * 60)
print("PHASE 1: TRAINING CLASSIFIER HEAD")
print("=" * 60)


# Freeze all feature layers

for param in model.features.parameters():

    param.requires_grad = False


# Make classifier trainable

for param in model.classifier.parameters():

    param.requires_grad = True


# ============================================================
# STEP 10: Phase 1 Optimizer and Loss
# ============================================================

optimizer = torch.optim.Adam(

    filter(
        lambda p: p.requires_grad,
        model.parameters()
    ),

    lr=1e-4
)


criterion = nn.CrossEntropyLoss()


# ============================================================
# STEP 11: Training Function
# ============================================================

def train_epoch(
    model,
    loader,
    optimizer,
    criterion,
    device
):

    model.train()

    total_loss = 0

    correct = 0

    total = 0


    for images, labels in tqdm(
        loader,
        desc='Training'
    ):

        images = images.to(device)

        labels = labels.to(device)


        optimizer.zero_grad()


        outputs = model(images)


        loss = criterion(
            outputs,
            labels
        )


        loss.backward()


        optimizer.step()


        total_loss += loss.item()


        _, predicted = outputs.max(1)


        correct += (
            predicted.eq(labels)
            .sum()
            .item()
        )


        total += labels.size(0)


    return (
        total_loss / len(loader),
        correct / total
    )


# ============================================================
# STEP 12: Evaluation Function
# ============================================================

def evaluate(
    model,
    loader,
    criterion,
    device
):

    model.eval()

    total_loss = 0

    correct = 0

    total = 0


    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)

            labels = labels.to(device)


            outputs = model(images)


            loss = criterion(
                outputs,
                labels
            )


            total_loss += loss.item()


            _, predicted = outputs.max(1)


            correct += (
                predicted.eq(labels)
                .sum()
                .item()
            )


            total += labels.size(0)


    return (
        total_loss / len(loader),
        correct / total
    )


# ============================================================
# STEP 13: Phase 1 Training
# ============================================================

train_losses = []

train_accs = []

val_losses = []

val_accs = []


best_val_acc = 0

patience_counter = 0

PATIENCE = 5


for epoch in range(30):

    train_loss, train_acc = train_epoch(

        model,

        train_loader,

        optimizer,

        criterion,

        device

    )


    val_loss, val_acc = evaluate(

        model,

        test_loader,

        criterion,

        device

    )


    train_losses.append(
        train_loss
    )

    train_accs.append(
        train_acc
    )

    val_losses.append(
        val_loss
    )

    val_accs.append(
        val_acc
    )


    print(

        f"Epoch {epoch+1}/30 | "

        f"Train Loss: {train_loss:.4f} | "

        f"Train Acc: {train_acc*100:.2f}% | "

        f"Val Loss: {val_loss:.4f} | "

        f"Val Acc: {val_acc*100:.2f}%"

    )


    # --------------------------------------------------------
    # Save best Phase 1 model
    # --------------------------------------------------------

    if val_acc > best_val_acc:

        best_val_acc = val_acc

        torch.save(

            model.state_dict(),

            '/content/drive/MyDrive/'
            'best_convnext_phase1.pth'

        )

        patience_counter = 0


    else:

        patience_counter += 1


        if patience_counter >= PATIENCE:

            print(
                "Early stopping!"
            )

            break


print(

    f"\nBest Phase 1 Validation Accuracy: "

    f"{best_val_acc*100:.2f}%"

)


# ============================================================
# STEP 14: Verify Phase 1 Checkpoint
# ============================================================

phase1_path = (
    '/content/drive/MyDrive/'
    'best_convnext_phase1.pth'
)


import os


print(
    "\nPhase 1 checkpoint exists:",
    os.path.exists(phase1_path)
)


# ============================================================
# STEP 15: PHASE 2 - FINE-TUNING
# ============================================================

print("\n")
print("=" * 60)
print("PHASE 2: FINE-TUNING")
print("=" * 60)


# ============================================================
# STEP 16: Load Best Phase 1 Model
# ============================================================

model.load_state_dict(

    torch.load(
        phase1_path,
        map_location=device
    )

)


print(
    "Best Phase 1 model loaded successfully."
)


# ============================================================
# STEP 17: Freeze Everything
# ============================================================

for param in model.parameters():

    param.requires_grad = False


# ============================================================
# STEP 18: Unfreeze Classifier
# ============================================================

for param in model.classifier.parameters():

    param.requires_grad = True


# ============================================================
# STEP 19: Unfreeze Last ConvNeXt Stage
#
# ConvNeXt-Tiny has four main feature stages.
# model.features[-1] is the final feature stage.
# ============================================================

for param in model.features[-1].parameters():

    param.requires_grad = True


# ============================================================
# STEP 20: Check Trainable Parameters
# ============================================================

print("\nTrainable parameters:")


trainable_params = 0

total_params = 0


for name, param in model.named_parameters():

    total_params += param.numel()


    if param.requires_grad:

        trainable_params += param.numel()

        print(name)


print(
    f"\nTrainable parameters: "
    f"{trainable_params:,}"
)


print(
    f"Total parameters: "
    f"{total_params:,}"
)


print(

    f"Percentage trainable: "

    f"{100 * trainable_params / total_params:.2f}%"

)


# ============================================================
# STEP 21: Phase 2 Optimizer
# ============================================================

optimizer_fine = torch.optim.Adam(

    filter(
        lambda p: p.requires_grad,
        model.parameters()
    ),

    lr=1e-5
)


criterion = nn.CrossEntropyLoss()


# ============================================================
# STEP 22: Phase 2 Training
# ============================================================

best_val_acc_fine = 0

patience_counter = 0

PATIENCE = 5


for epoch in range(20):

    train_loss, train_acc = train_epoch(

        model,

        train_loader,

        optimizer_fine,

        criterion,

        device

    )


    val_loss, val_acc = evaluate(

        model,

        test_loader,

        criterion,

        device

    )


    train_losses.append(
        train_loss
    )

    train_accs.append(
        train_acc
    )

    val_losses.append(
        val_loss
    )

    val_accs.append(
        val_acc
    )


    print(

        f"Fine-tuning Epoch {epoch+1}/20 | "

        f"Train Loss: {train_loss:.4f} | "

        f"Train Acc: {train_acc*100:.2f}% | "

        f"Val Loss: {val_loss:.4f} | "

        f"Val Acc: {val_acc*100:.2f}%"

    )


    # --------------------------------------------------------
    # Save Best Phase 2 Model
    # --------------------------------------------------------

    if val_acc > best_val_acc_fine:

        best_val_acc_fine = val_acc


        torch.save(

            model.state_dict(),

            '/content/drive/MyDrive/'
            'best_convnext_fine.pth'

        )


        patience_counter = 0


    else:

        patience_counter += 1


        if patience_counter >= PATIENCE:

            print(
                "Early stopping!"
            )

            break


print(

    f"\nBest Phase 2 Validation Accuracy: "

    f"{best_val_acc_fine*100:.2f}%"

)


# ============================================================
# STEP 23: Load Best Fine-Tuned Model
# ============================================================

phase2_path = (
    '/content/drive/MyDrive/'
    'best_convnext_fine.pth'
)


model.load_state_dict(

    torch.load(
        phase2_path,
        map_location=device
    )

)


model = model.to(device)


print(
    "\nBest fine-tuned ConvNeXt model "
    "loaded successfully."
)


# ============================================================
# STEP 24: Plot Training History
# ============================================================

fig, (ax1, ax2) = plt.subplots(

    1,
    2,
    figsize=(12, 4)

)


# Accuracy

ax1.plot(
    train_accs,
    label='Train Accuracy'
)

ax1.plot(
    val_accs,
    label='Val Accuracy'
)

ax1.set_title(
    'ConvNeXt Accuracy'
)

ax1.set_xlabel(
    'Epoch'
)

ax1.set_ylabel(
    'Accuracy'
)

ax1.legend()


# Loss

ax2.plot(
    train_losses,
    label='Train Loss'
)

ax2.plot(
    val_losses,
    label='Val Loss'
)

ax2.set_title(
    'ConvNeXt Loss'
)

ax2.set_xlabel(
    'Epoch'
)

ax2.set_ylabel(
    'Loss'
)

ax2.legend()


plt.tight_layout()


plt.savefig(

    '/content/drive/MyDrive/'
    'convnext_accuracy_loss.png'

)


plt.show()


# ============================================================
# STEP 25: Classification Report
# ============================================================

model.eval()


all_preds = []

all_labels = []


with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)


        outputs = model(images)


        _, predicted = outputs.max(1)


        all_preds.extend(
            predicted.cpu().numpy()
        )


        all_labels.extend(
            labels.numpy()
        )


# ============================================================
# Test Accuracy
# ============================================================

test_accuracy = (

    (
        np.array(all_preds)
        ==
        np.array(all_labels)
    ).mean()

    * 100

)


print(
    f"\nTest Accuracy: "
    f"{test_accuracy:.2f}%"
)


# ============================================================
# Classification Report
# ============================================================

print(
    "\nClassification Report:"
)


print(

    classification_report(

        all_labels,

        all_preds,

        target_names=train_dataset.classes

    )

)


# ============================================================
# STEP 26: Confusion Matrix
# ============================================================

cm = confusion_matrix(

    all_labels,

    all_preds

)


plt.figure(

    figsize=(20, 16)

)


sns.heatmap(

    cm,

    annot=True,

    fmt='d',

    xticklabels=train_dataset.classes,

    yticklabels=train_dataset.classes

)


plt.title(
    'Confusion Matrix - ConvNeXt'
)

plt.ylabel(
    'True Label'
)

plt.xlabel(
    'Predicted Label'
)

plt.xticks(
    rotation=90
)

plt.tight_layout()


plt.savefig(

    '/content/drive/MyDrive/'
    'convnext_confusion_matrix.png'

)


plt.show()


# ============================================================
# STEP 27: Save Final Model
# ============================================================

final_model_path = (

    '/content/drive/MyDrive/'
    'convnext_plantdoc.pth'

)


torch.save(

    model.state_dict(),

    final_model_path

)


print(
    "\n============================================================"
)

print(
    "ConvNeXt model saved successfully!"
)

print(
    f"Location: {final_model_path}"
)

print(
    "============================================================"
)
```
